Here we will review all the methods for handling missing values during preprocessing in the machine learning process.

```python
import numpy as np
import pandas as pd
```

```python
# Loading Data
data = pd.read_csv('Loan_Default.csv')
```

```python
# Identifying the number of empty cells in all columns of our dataset 
data.isna().sum()
```

# Method 1: Simple deletion of missing values in our dataset

```python
data_suppress = data.dropna()
# Verification
data_suppress.isna().sum()
```

# Method 2: Separating categorical and numerical data
Categorical variables

```python
data_cat = data.select_dtypes(include=['object'])
# Variables numériques
data_num = data.select_dtypes(include=['float64','int64'])
```

```python
# Imputation of arbitrary values for missing data in categorical and numerical variables using a constant 

data_cat.fillna('Unknown',inplace=True)
data_num.fillna(0,inplace=True)
```

```python
data_imputation = pd.concat([data_cat,data_num],axis=1)
# Verification of all data  
data_imputation.isna().sum()
```

# Method 3: Assigning arbitrary values to missing data in categorical and numerical variables using a statistic

```python
data_imputation_stat = data.copy()
```

```python
#For numerical variables, replace the empty cells in the column with the column mean
data_imputation_stat['term'] = data_imputation_stat['term'].fillna(data_imputation_stat['term'].mean())
data_imputation_stat['property_value'] = data_imputation_stat['property_value'].fillna(data_imputation_stat['property_value'].mean())
data_imputation_stat['income'] = data_imputation_stat['income'].fillna(data_imputation_stat['income'].mean())
data_imputation_stat['LTV'] = data_imputation_stat['LTV'].fillna(data_imputation_stat['LTV'].mean())
data_imputation_stat['dtir1'] = data_imputation_stat['dtir1'].fillna(data_imputation_stat['dtir1'].mean())
data_imputation_stat['rate_of_interest'] = data_imputation_stat['rate_of_interest'].fillna(data_imputation_stat['rate_of_interest'].mean())
data_imputation_stat['Interest_rate_spread'] = data_imputation_stat['Interest_rate_spread'].fillna(data_imputation_stat['Interest_rate_spread'].mean())
data_imputation_stat['Upfront_charges'] = data_imputation_stat['Upfront_charges'].fillna(data_imputation_stat['Upfront_charges'].mean())


# For categorical variables, replace empty cells in the column with the most frequently occurring element in that column
data_imputation_stat['Neg_ammortization'] = data_imputation_stat['Neg_ammortization'].fillna(data_imputation_stat['Neg_ammortization'].mode()[0])
data_imputation_stat['age'] = data_imputation_stat['age'].fillna(data_imputation_stat['age'].mode()[0])
data_imputation_stat['submission_of_application'] = data_imputation_stat['submission_of_application'].fillna(data_imputation_stat['submission_of_application'].mode()[0])
data_imputation_stat['loan_limit'] = data_imputation_stat['loan_limit'].fillna(data_imputation_stat['loan_limit'].mode()[0])
data_imputation_stat['approv_in_adv'] = data_imputation_stat['approv_in_adv'].fillna(data_imputation_stat['approv_in_adv'].mode()[0])
data_imputation_stat['loan_purpose'] = data_imputation_stat['loan_purpose'].fillna(data_imputation_stat['loan_purpose'].mode()[0])


# Check the entire dataset 
data_imputation_stat.isna().sum()
```

# Method 5: Implementing the nearest neighbor method to compensate for missing values in our dataset (KNN method)
This method applies exclusively to numerical data

```python
data_knn_imput = data.copy()
```

```python
data_num_2 = data_knn_imput.select_dtypes(include=['float64','int64'])
data_cat_2 = data_knn_imput.select_dtypes(include=['object'])
```

```python
data_num_2 = list(data_num_2)
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.impute import KNNImputer

scaler = StandardScaler()
data_knn_num = data[data_num_2]
data_knn_num_scaled = scaler.fit_transform(data_knn_num)

knn_imputer = KNNImputer(n_neighbors=5) 
knn_imputer.fit(data_knn_num_scaled)
 
data_knn_num_imputed = knn_imputer.fit_transform(data_knn_num_scaled)
```

```python
# Verifying the removal of empty cells in the numerical columns of our dataset 
data_knn_num_final = pd.DataFrame(data_knn_num_imputed, columns=data_num_2)
data_knn_num_final.isna().sum()
```
