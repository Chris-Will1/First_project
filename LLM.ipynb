Large Langage Models (LLM) are type of generative AI that specializes in generating coherent responses to input text.
N.B : LLM is considered as Deep Learning based upon massive corpus of text. 
It can be used for a wide range of natural langage processing task.
Example --> LLM allows us to have specialized chatbot 
How it works :
LLMs takes an input text (known as prompt) and break the input into sequence of numerical tokens (tokenization), the model then uses a probability distribution to predict the next token and repeat the same method for each token.

```python
# First example of LLM usage
# We will use the Mistral 7B model

from transformers import AutoTokenizer, AutoModelForCausalLM

MODEL = "mistralai/Mistral-7B-Instruct-v0.2"

tokenizer = AutoTokenizer.from_pretrained(MODEL)

model = AutoModelForCausalLM.from_pretrained(
    MODEL,
    device_map="auto",
    torch_dtype="auto"
)

prompt = "Explique en termes simples ce qu'est l'intelligence artificielle :"

# Break the prompt into a sequence of numerical tokens 
inputs = tokenizer(prompt, return_tensors="pt").to(model.device)

# Creation of the response to the initial prompt
# The outputs will take the prompt (tokenized) and ask the model (mistral7b) to generate an accurate and creative response

outputs = model.generate(
    **inputs,
    max_new_tokens=200,
    temperature=0.7,
    top_p=0.9
)

# Displaying the response to our prompt
print(tokenizer.decode(outputs[0], skip_special_tokens=True))
```

```python
import torch
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

# Here we take the FLAN-T5 base model

MODEL = "google/flan-t5-base"
device = "mps" if torch.backends.mps.is_available() else ("cuda" if torch.cuda.is_available() else "cpu")

# 1st step : take the input (a prompt)
tokenizer = AutoTokenizer.from_pretrained(MODEL)

# This step will take the original prompt into the model
model = AutoModelForSeq2SeqLM.from_pretrained(MODEL).to(device)
 
prompt = "Explique en termes simples ce qu'est l'intelligence artificielle :"

# break the input into sequence of numerical tokens (tokenization)
inputs = tokenizer(prompt, return_tensors="pt").to(device)

# the model then uses a probability distribution to predict the next token and repeat the same method for each token
outputs = model.generate(**inputs, max_new_tokens=150)

# Displaying the response to our prompt
print(tokenizer.decode(outputs[0], skip_special_tokens=True))
```

```python
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM

MODEL = "facebook/opt-350m"


device = "mps" if torch.backends.mps.is_available() else "cpu"
print("Using device:", device)

# Tokenizer
tokenizer = AutoTokenizer.from_pretrained(MODEL)

model = AutoModelForCausalLM.from_pretrained(
    MODEL,
    dtype=torch.float16  # utiliser dtype au lieu de torch_dtype
).to(device)

# Prompt
prompt = "Explique en termes simples la différence entre le machine learning supervisé et non supervisé."
inputs = tokenizer(prompt, return_tensors="pt").to(device)

# Generation of the optimized response
outputs = model.generate(
    input_ids=inputs["input_ids"],
    attention_mask=inputs["attention_mask"],
    max_new_tokens=150,
    temperature=0.7,
    top_p=0.9,
    do_sample=True
)

# Displaying the response to our prompt
print(tokenizer.decode(outputs[0], skip_special_tokens=True, clean_up_tokenization_spaces=True))
```

```python
# BERT is bidirectional and widely used for comprehension tasks (not generation).
# For comprehension, classification, extraction

from transformers import pipeline

# BERT fine-tuned for sentiment classification
classifier = pipeline("sentiment-analysis", model="nlptown/bert-base-multilingual-uncased-sentiment")

texte = "J'adore ce film, il est incroyable !"
result = classifier(texte)
print(result)
```

Our model: nlptown/bert-base-multilingual-uncased-sentiment
Our model has been trained to classify the sentiment of sentences on a scale of 1 to 5 stars.
1 star = very negative
2 stars = negative
3 stars = neutral/average
4 stars = positive
5 stars = very positive

```python
# T5 is an encoder-decoder (seq2seq) model, perfect for translation, summarization, and Q&A.
# For text transformation (summarization, translation, Q&A).

from transformers import pipeline

# Load a T5 model that has already been fine-tuned for summarization
summarizer = pipeline("summarization", model="t5-small")

# Example text

texte = """
L'intelligence artificielle (IA) désigne l'ensemble des théories et des techniques développant
des programmes informatiques capables de simuler des traits de l'intelligence humaine tels que
le raisonnement, l'apprentissage, la planification ou encore la créativité.
"""

res = summarizer(
    texte,
    max_new_tokens=60,   # maximum length of the generated summary
    min_length=20,       # summary of at least 20
    do_sample=False      # deterministic generation
)

print("Résumé :", res)

# The summary of our prompt will be very similar to our initial prompt because the latter is not very long
# The idea here is to understand how rapid synthesis works and how to implement an LLM for this purpose.
```

```python
# Same model in a more interpretable form for possible presentations to teams unfamiliar with LLM methods 

from transformers import pipeline

# Loading the pipeline
classifier = pipeline("sentiment-analysis", model="nlptown/bert-base-multilingual-uncased-sentiment")

# Correspondence dictionary
sentiment_map = {
    "1 star": "Très négatif",
    "2 stars": "Négatif",
    "3 stars": "Neutre",
    "4 stars": "Positif",
    "5 stars": "Très positif"
}

def analyze_sentiment(text):
    result = classifier(text)[0]
    label = result['label']
    score = result['score']
    sentiment = sentiment_map.get(label, label)
    return {"sentiment": sentiment, "score": round(score, 3)}

# Tests
print(analyze_sentiment("J'adore ce film, il est incroyable !"))
print(analyze_sentiment("C'est correct, sans plus."))
print(analyze_sentiment("Je déteste ce produit, il est nul !"))
```

Practical application of LLM principles for sentiment classification on a dataset

```python
import pandas as pd

data = pd.read_csv(
    'product_reviews.csv',
    on_bad_lines='skip',
    sep=';'
)

display(data)
```

```python
data.drop(['Unnamed: 2','Unnamed: 3','Unnamed: 4'], axis=1, inplace=True)
data.head()
```

```python
from transformers import pipeline

# BERT fine-tuned for sentiment classification
classifier = pipeline("sentiment-analysis", model="nlptown/bert-base-multilingual-uncased-sentiment")

texte = data['text']
results = []
def prediction (text):
    for texte in data['text']:
        result = classifier(texte)
        results.append(result)
    #print(result)
    #print(texte)
data['prediction'] = data['text'].apply(prediction)
data.head()
```

```python
# Algorithm for summarizing a prompt 

import pandas as pd
from transformers import pipeline

summarizer = pipeline("summarization", model="facebook/bart-large-cnn")  # modèle open-access

# Function to summarize a text
def summarize_text(text, max_length=None):
    #  Automatic adjustment of max_length if necessary
    if max_length is None:
        input_len = len(text.split())
        max_length = max(10, min(50, input_len // 2))
    
    summary = summarizer(text, max_length=max_length, min_length=5, do_sample=False)
    return summary[0]["summary_text"]

# Apply to the column
data["resume"] = data["text"].apply(summarize_text)
data
```

```python
def predict_sentiment(text):
    result = classifier(text)[0]  
    return result['label']
data['prediction'] = data['text'].apply(predict_sentiment)
data
```

```python
# Addition of a column for improved interpretability 
def libelle(text):
    if text == '1 star':
        return 'Très mauvais'
    elif text == '2 stars':
        return 'Mauvais'
    elif text == '3 stars':
        return 'Neutre'
    elif text == '4 stars':
        return 'Bon'
    else:
        return 'Très bon'
data['Libellé']=data['prediction'].apply(libelle)
data
```
