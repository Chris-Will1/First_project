Here we will focus on two data visualization libraries (Matplotlib and Seaborn) and how to use them.

# Matplotlib

```python
# Simple Curve 
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

plt.plot(x, y, marker="o", color="black", label="y=2x")
plt.xlabel("X")
plt.ylabel("Y")
plt.legend()
plt.show()
```

```python
# Simple histogram
import matplotlib.pyplot as plt
import numpy as np

data = np.random.randn(1000)

plt.hist(data, bins=20, color="skyblue", edgecolor="black")
plt.title("Histogram - data distribution")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.show()
```

```python
# histogram with multiple distributions 

data1 = np.random.normal(0, 1, 1000)   # moyenne 0, écart-type 1
data2 = np.random.normal(3, 1.5, 1000) # moyenne 3, écart-type 1.5
data3 = np.random.normal(6, 2, 1000)   # moyenne 6, écart-type 2
data4 = np.random.normal(9, 2.5, 1000) # moyenne 9, écart-type 2.5

plt.hist(data1, bins=30, alpha=1, label="Dist 1")
plt.hist(data2, bins=30, alpha=1, label="Dist 2")
plt.hist(data3, bins=30, alpha=1, label="Dist 3")
plt.hist(data4, bins=30, alpha=1, label="Dist 4")
plt.title("Comparative histogram")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.legend()
plt.show()
```

```python
# Bar chart 

import matplotlib.pyplot as plt

categories = ["A", "B", "C", "D", "E"]
valeurs = [5, 7, 3, 8, 6]

plt.bar(categories, valeurs, color="blue", edgecolor="black")
plt.title("Bar chart")
plt.xlabel("Categories")
plt.ylabel("Values")
plt.show()
```

```python
# Horizontal bar 

plt.barh(categories, valeurs, color="lightgreen", edgecolor="black")
plt.title("Horizontal bar chart")
plt.xlabel("Values")
plt.ylabel("Categories")
plt.show()
```

```python
# Comparisons of 2 series  
import numpy as np

# Données
categories = ["A", "B", "C", "D"]
valeurs1 = [4, 7, 3, 5]
valeurs2 = [6, 2, 5, 8]

x = np.arange(len(categories))  # position des barres
largeur = 0.35  # largeur des barres

# Bar chart comparatif
plt.bar(x - largeur/2, valeurs1, largeur, label="Série 1", color="skyblue")
plt.bar(x + largeur/2, valeurs2, largeur, label="Série 2", color="orange")

plt.xticks(x, categories)
plt.title("Comparisons of 2 series")
plt.xlabel("Categories")
plt.ylabel("Values")
plt.legend()
plt.show()
```

```python
# Data scatter plots 

import matplotlib.pyplot as plt
import numpy as np

x = np.random.rand(50)
y = 2 * x + np.random.randn(50) * 0.1  

plt.scatter(x, y, marker="o", color="black")
plt.title("Simple scatter plots")
plt.xlabel("Variable X")
plt.ylabel("Variable Y")
plt.show()
```

```python
# Enhanced scatter plot

x = np.random.rand(100)
y = np.random.rand(100)
taille = np.random.rand(100) * 300   
couleur = np.random.rand(100)       

plt.scatter(x, y, s=taille,c=couleur, cmap="viridis", alpha=0.7, edgecolors="black")
plt.title("Scatter plot with sizes and colors")
plt.xlabel("X")
plt.ylabel("Y")
plt.colorbar(label="Related values") 
plt.show()
```

```python
# Several groups of scatter plots

# Group 1
x1 = np.random.normal(2, 0.5, 50)
y1 = np.random.normal(2, 0.5, 50)

# Group 2
x2 = np.random.normal(4, 0.5, 50)
y2 = np.random.normal(4, 0.5, 50)

plt.scatter(x1, y1, color="red", label="Group 1", alpha=0.7)
plt.scatter(x2, y2, color="blue", label="Group 2", alpha=0.7)
plt.title("Point clouds by groups")
plt.xlabel("X")
plt.ylabel("Y")
plt.legend()
plt.show()
```

# Seaborn 

```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# Simple Dataset
np.random.seed(42)
df = pd.DataFrame({
    "x": np.arange(1, 21),
    "y": np.random.randint(10, 100, 20),
    "cat": np.random.choice(["A", "B", "C"], size=20),
    "values": np.random.randn(20)
})

sns.set_theme(style="whitegrid")
```

```python
# Examples of different styles 
for style in ["darkgrid", "whitegrid", "dark", "white", "ticks"]:
    sns.set_theme(style=style)
    sns.histplot(np.random.randn(100), bins=20)
    plt.title(f"Style = {style}")
    plt.show()
```

```python
# Curves 
sns.lineplot(x="x",y="y",data=df, marker="o", hue="cat")
plt.title("Curve with Seaborn (lineplot)")
plt.show()
```

```python
# Histograms
sns.histplot(df["values"], bins = 20, kde=True, color='grey', edgecolor='black')
plt.title("Histogram with Seaborn (histplot)")
plt.show()
```

```python
# Barre verticale 
sns.barplot(x="cat", y="y", data=df, estimator=np.mean, palette="viridis")
plt.title("Diagramme en barres verticales")
plt.show()
```
